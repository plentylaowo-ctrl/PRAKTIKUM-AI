```python
# Bagian 1: Data Diri

# Mendefinisikan variabel untuk data diri
nama_saya = "Plenty Elegance Laowo"
umur_saya = 20
tinggi_saya = 150.0
is_mahasiswa = True

# Menampilkan data diri
print(f"Nama: {nama_saya}, Umur: {umur_saya} tahun, Tinggi: {tinggi_saya} cm, Status: Mahasiswa")

# Menampilkan tipe data dari setiap variabel
print(f"Tipe data nama_saya: {type(nama_saya)}")
print(f"Tipe data umur_saya: {type(umur_saya)}")
print(f"Tipe data tinggi_saya: {type(tinggi_saya)}")
print(f"Tipe data is_mahasiswa: {type(is_mahasiswa)}")

print("-" * 30)
```

```python
# Bagian 2: Data 5 Teman

# Mendefinisikan empat list untuk data 5 teman
nama_teman = ["Andi", "Budi", "Citra", "Dina", "Eko"]
umur_teman = [20, 19, 21, 20, 22]
hobi_teman = ["Membaca", "Sepak Bola", "Menyanyi", "Memasak", "Gaming"]
kota_asal = ["Jakarta", "Bandung", "Surabaya", "Medan", "Yogyakarta"]

# Menampilkan panjang dan tipe data dari setiap list
print("Panjang setiap list:")
print(f"Panjang list nama_teman: {len(nama_teman)}")
print(f"Panjang list umur_teman: {len(umur_teman)}")
print(f"Panjang list hobi_teman: {len(hobi_teman)}")
print(f"Panjang list kota_asal: {len(kota_asal)}")

print("\nTipe data setiap list:")
print(f"Tipe data nama_teman: {type(nama_teman)}")
print(f"Tipe data umur_teman: {type(umur_teman)}")
print(f"Tipe data hobi_teman: {type(hobi_teman)}")
print(f"Tipe data kota_asal: {type(kota_asal)}")

print("-" * 30)
```

```python
# Bagian 3: Menampilkan Data Teman dengan Loop dan Manual

# Menggunakan loop for untuk menampilkan data semua teman
print("Menampilkan data semua teman dengan loop:")
for i in range(len(nama_teman)):
    print(f"Teman ke-{i+1}: {nama_teman[i]}, Umur: {umur_teman[i]} tahun, Hobi: {hobi_teman[i]}, Asal: {kota_asal[i]}")

print("-" * 30)

# Menampilkan data menggunakan indexing manual
print("Menampilkan data teman menggunakan indexing manual:")
# Data teman pertama
print(f"Data teman pertama: {nama_teman[0]}, Umur: {umur_teman[0]} tahun, Hobi: {hobi_teman[0]}, Asal: {kota_asal[0]}")
# Data teman terakhir
print(f"Data teman terakhir: {nama_teman[-1]}, Umur: {umur_teman[-1]} tahun, Hobi: {hobi_teman[-1]}, Asal: {kota_asal[-1]}")

print("-" * 30)

# Menampilkan sebagian data menggunakan slicing
print("Menampilkan sebagian data menggunakan slicing:")
# Tampilkan 3 nama teman pertama
print(f"Nama 3 teman pertama: {nama_teman[0:3]}")
```
